## Data

```python
import numpy as np

# Inputs & targets
x1, x2 = 0.05, 0.10
T1, T2 = 0.01, 0.99

# Weights: input→hidden
w1, w2, w3, w4 = 0.15, 0.20, 0.25, 0.30
# Weights: hidden→output
w5, w6, w7, w8 = 0.40, 0.45, 0.50, 0.55

# Biases
b1, b2 = 0.35, 0.60

lr = 0.5  # learning rate
```

# Forward Pass

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

# Hidden layer net input
net_H1 = w1*x1 + w3*x2 + b1
net_H2 = w2*x1 + w4*x2 + b1

H1 = sigmoid(net_H1)
H2 = sigmoid(net_H2)

# Output layer net input
net_y1 = w5*H1 + w7*H2 + b2
net_y2 = w6*H1 + w8*H2 + b2

y1 = sigmoid(net_y1)
y2 = sigmoid(net_y2)

print(f"H1={H1:.4f}, H2={H2:.4f}")
print(f"y1={y1:.4f}, y2={y2:.4f}")
```

# Loss (MSE)

```python
# Total error = sum of 0.5*(target - output)^2
E1 = 0.5 * (T1 - y1)**2
E2 = 0.5 * (T2 - y2)**2
E_total = E1 + E2

print(f"Total Error: {E_total:.6f}")
```

# Backpropagation

```python
# --- Output layer deltas ---
# δE/δy * δy/δnet  (sigmoid derivative = y*(1-y))
delta_y1 = (y1 - T1) * y1 * (1 - y1)
delta_y2 = (y2 - T2) * y2 * (1 - y2)

# --- Gradients for hidden→output weights ---
dw5 = delta_y1 * H1
dw6 = delta_y2 * H1
dw7 = delta_y1 * H2
dw8 = delta_y2 * H2

# --- Hidden layer deltas (error flows back through both outputs) ---
delta_H1 = (delta_y1*w5 + delta_y2*w6) * H1 * (1 - H1)
delta_H2 = (delta_y1*w7 + delta_y2*w8) * H2 * (1 - H2)

# --- Gradients for input→hidden weights ---
dw1 = delta_H1 * x1
dw2 = delta_H2 * x1
dw3 = delta_H1 * x2
dw4 = delta_H2 * x2
```

# Weight Update

```python
w1 -= lr * dw1
w2 -= lr * dw2
w3 -= lr * dw3
w4 -= lr * dw4
w5 -= lr * dw5
w6 -= lr * dw6
w7 -= lr * dw7
w8 -= lr * dw8

print(f"Updated w1={w1:.4f}, w5={w5:.4f}")
print(f"Updated w2={w2:.4f}, w6={w6:.4f}")
```

# Full Training Loop

```python
# Reset weights to original values
w1,w2,w3,w4 = 0.15,0.20,0.25,0.30
w5,w6,w7,w8 = 0.40,0.45,0.50,0.55

for epoch in range(10000):
    # Forward
    H1 = sigmoid(w1*x1 + w3*x2 + b1)
    H2 = sigmoid(w2*x1 + w4*x2 + b1)
    y1 = sigmoid(w5*H1 + w7*H2 + b2)
    y2 = sigmoid(w6*H1 + w8*H2 + b2)

    # Backward
    d1 = (y1-T1)*y1*(1-y1)
    d2 = (y2-T2)*y2*(1-y2)
    dH1 = (d1*w5 + d2*w6)*H1*(1-H1)
    dH2 = (d1*w7 + d2*w8)*H2*(1-H2)

    # Update
    w5 -= lr*d1*H1;  w6 -= lr*d2*H1
    w7 -= lr*d1*H2;  w8 -= lr*d2*H2
    w1 -= lr*dH1*x1; w2 -= lr*dH2*x1
    w3 -= lr*dH1*x2; w4 -= lr*dH2*x2

    if epoch % 1000 == 0:
        loss = 0.5*((T1-y1)**2 + (T2-y2)**2)
        print(f"Epoch {epoch}: Loss={loss:.6f} | y1={y1:.4f}, y2={y2:.4f}")
```
